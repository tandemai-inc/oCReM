# oCReM Structure Generation Tutorial

This tutorial demonstrates structure generation capabilities of oCReM (optimized Chemical Reassembly), including:
- Using fragment databases to generate new molecular structures

Here I used the sqlite3 database created in the tutorial, 'fragmentation_to_db.ipynb'.

## Mutate molecule <a name='mutate'></a>




```python
import os
import sys
from rdkit import Chem

sys.path.append(os.path.dirname(os.getcwd())) 

from ta_gen.ocrem.ocrem import mutate_mol
from ta_gen.db import create_db_manager

m = Chem.MolFromSmiles('OCCOc1ccccc1')  # toluene
db_manager = create_db_manager('sqlite', db_path='replacements.db')
# db_manager = create_db_manager('postgres', ini_file='replacements.ini')
mols = list(mutate_mol(m, db_manager, max_inc=1))

mols
```

## Grow molecule <a name='grow'></a>

```python
import os
import sys
from rdkit import Chem

sys.path.append(os.path.dirname(os.getcwd())) 

from ta_gen.ocrem.ocrem import grow_mol
from ta_gen.db import create_db_manager

m = Chem.MolFromSmiles('OCCOc1ccccc1')  # toluene
db_manager = create_db_manager('sqlite', db_path='replacements.db')
# db_manager = create_db_manager('postgres', ini_file='replacements.ini')
mols = list(grow_mol(m, db_manager))

mols
```

## Link molecule <a name='grow'></a>

```python
import os
import sys
from rdkit import Chem

sys.path.append(os.path.dirname(os.getcwd())) 

from ta_gen.ocrem.ocrem import link_mols
from ta_gen.db import create_db_manager

m1 = Chem.MolFromSmiles('OC=O')  # toluene
m2 = Chem.MolFromSmiles('CC=O')  # glycine
db_manager = create_db_manager('sqlite', db_path='replacements.db')
# db_manager = create_db_manager('postgres', ini_file='replacements.ini')
mols = list(link_mols(m1, m2, db_manager, radius=1))

mols
```
